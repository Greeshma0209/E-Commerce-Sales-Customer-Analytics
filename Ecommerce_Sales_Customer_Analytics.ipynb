```python
total_sales = df_clean['total_amount'].sum()

total_orders = df_clean['order_id'].nunique()

total_customers = df_clean['customer_id'].nunique()

average_order_value = df_clean['total_amount'].mean()

print("Total Sales:", total_sales)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Average Order Value:", average_order_value)
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("Desktop/E-Commerce Sales & Customer Analytics/Raw_Data/ecommerce_sales.csv")

df.head()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])

df_clean = df.copy()

df_clean['year'] = df_clean['order_date'].dt.year
df_clean['month'] = df_clean['order_date'].dt.month
df_clean['month_name'] = df_clean['order_date'].dt.month_name()
```

```python
df_clean.head()
```

```python
monthly_sales = df_clean.groupby(
    ['year', 'month']
)['total_amount'].sum().reset_index()

monthly_sales['year_month'] = (
    monthly_sales['year'].astype(str)
    + '-'
    + monthly_sales['month'].astype(str).str.zfill(2)
)

monthly_sales.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales['year_month'],
    monthly_sales['total_amount'],
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_customers = df_clean.groupby('customer_id').agg(
    Total_Spent=('total_amount', 'sum'),
    Total_Orders=('order_id', 'count')
).sort_values(by='Total_Spent', ascending=False).head(10)

top_customers
```

```python
customer_sales = df_clean.groupby('customer_id').agg(
    Total_Spent=('total_amount', 'sum'),
    Total_Orders=('order_id', 'count')
).reset_index()

customer_sales.head()
```

```python
customer_sales['Customer_Segment'] = pd.qcut(
    customer_sales['Total_Spent'],
    q=3,
    labels=['Low Value', 'Medium Value', 'High Value']
)

customer_sales.head()
```

```python
segment_analysis = customer_sales.groupby('Customer_Segment', observed=True).agg(
    Customers=('customer_id', 'count'),
    Total_Sales=('Total_Spent', 'sum'),
    Average_Spending=('Total_Spent', 'mean')
)

segment_analysis
```

```python
gender_analysis = df_clean.groupby('customer_gender').agg(
    Customers=('customer_id', 'nunique'),
    Total_Sales=('total_amount', 'sum'),
    Average_Spending=('total_amount', 'mean')
).sort_values(by='Total_Sales', ascending=False)

gender_analysis
```

```python
payment_analysis = df_clean.groupby('payment_method').agg(
    Total_Orders=('order_id', 'count'),
    Total_Sales=('total_amount', 'sum'),
    Average_Order_Value=('total_amount', 'mean')
).sort_values(by='Total_Sales', ascending=False)

payment_analysis
```

```python
product_analysis = df_clean.groupby('product_id').agg(
    Total_Sales=('total_amount', 'sum'),
    Total_Orders=('order_id', 'count'),
    Total_Quantity=('quantity', 'sum')
).sort_values(by='Total_Sales', ascending=False).head(10)

product_analysis
```

```python
low_products = df_clean.groupby('product_id').agg(
    Total_Sales=('total_amount', 'sum'),
    Total_Orders=('order_id', 'count'),
    Total_Quantity=('quantity', 'sum')
).sort_values(by='Total_Sales', ascending=True).head(10)

low_products
```

```python
return_analysis = df_clean.groupby('returned').agg(
    Total_Orders=('order_id', 'count'),
    Total_Sales=('total_amount', 'sum'),
    Average_Order_Value=('total_amount', 'mean')
)

return_analysis
```

```python
category_returns = df_clean.groupby('category').agg(
    Total_Orders=('order_id', 'count'),
    Returned_Orders=('returned', lambda x: (x == 'Yes').sum())
)

category_returns['Return_Rate'] = (
    category_returns['Returned_Orders']
    / category_returns['Total_Orders']
    * 100
)

category_returns.sort_values(
    by='Return_Rate',
    ascending=False
)
```

```python
customer_frequency = df_clean.groupby('customer_id').agg(
    Total_Orders=('order_id', 'count')
)

customer_frequency['Customer_Type'] = customer_frequency['Total_Orders'].apply(
    lambda x: 'One-time Customer' if x == 1 else 'Repeat Customer'
)

customer_frequency['Customer_Type'].value_counts()
```

```python
age_analysis = df_clean.groupby('customer_age').agg(
    Total_Customers=('customer_id', 'nunique'),
    Total_Sales=('total_amount', 'sum'),
    Average_Spending=('total_amount', 'mean')
).sort_values(by='Total_Sales', ascending=False)

age_analysis
```

```python
region_category = df_clean.groupby(
    ['region', 'category']
).agg(
    Total_Sales=('total_amount', 'sum'),
    Total_Orders=('order_id', 'count')
).sort_values(
    by=['region', 'Total_Sales'],
    ascending=[True, False]
)

region_category
```

```python
discount_analysis = df_clean.groupby('discount').agg(
    Total_Orders=('order_id', 'count'),
    Total_Sales=('total_amount', 'sum'),
    Average_Order_Value=('total_amount', 'mean')
).sort_values(by='discount')

discount_analysis
```

```python
quantity_analysis = df_clean.groupby('quantity').agg(
    Total_Orders=('order_id', 'count'),
    Total_Sales=('total_amount', 'sum'),
    Average_Order_Value=('total_amount', 'mean')
).sort_values(by='quantity')

quantity_analysis
```

```python
delivery_analysis = df_clean.groupby('delivery_time_days').agg(
    Total_Orders=('order_id', 'count'),
    Total_Sales=('total_amount', 'sum'),
    Average_Order_Value=('total_amount', 'mean')
).sort_values(by='delivery_time_days')

delivery_analysis
```

```python
shipping_analysis = df_clean.groupby('shipping_cost').agg(
    Total_Orders=('order_id', 'count'),
    Total_Sales=('total_amount', 'sum'),
    Average_Order_Value=('total_amount', 'mean')
).sort_values(by='shipping_cost')

shipping_analysis
```

```python
df_clean['shipping_cost_range'] = pd.cut(
    df_clean['shipping_cost'],
    bins=[-1, 5, 10, 15, 20],
    labels=['Low (₹0-5)', 'Medium (₹5-10)', 'High (₹10-15)', 'Very High (₹15-20)']
)

shipping_range_analysis = df_clean.groupby(
    'shipping_cost_range',
    observed=True
).agg(
    Total_Orders=('order_id', 'count'),
    Total_Sales=('total_amount', 'sum'),
    Average_Order_Value=('total_amount', 'mean')
)

shipping_range_analysis
```
